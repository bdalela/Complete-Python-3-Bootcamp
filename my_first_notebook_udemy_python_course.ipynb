```python
#if starts with a vowel- add 'ay' to the end
#if consonant, put first letter at end, then add 'ay'
#e.g. word--> ordway
#e.g. apple-->appleay
```

```python
def pig_latin(word):
    first_letter = word[0]
    #check if vowel
    if first_letter in 'aeiou':
        pig_word = word + 'ay'
    else:
        pig_word = word [1:]+ first_letter +'ay'
    return pig_word
```

```python
pig_latin('bharati')
```

```python
def myfunc(**kwargs):
    if 'fruit' in kwargs:
        print ("My fruit of choice is {}".format(kwargs['fruit']))
    else:
        print ("I didn't find any fruit here")
```

```python
myfunc(fruit='apple', veggie = 'lettuce')
```
